```python
def generate_filelistJF(pgroup, run):
    filelist = sorted(glob.glob('/sf/alvra/data/{}/res/processed/run{:04d}*/data/*acq*JF*'.format(pgroup, run)))
    return filelist

def generate_filelistJF_pp(pgroup, run):
    filelist = sorted(glob.glob('/sf/alvra/data/{}/res/processed/run{:04d}*/data/*acq*pp*'.format(pgroup, run)))
    return filelist

def generate_filelistJF_corr(pgroup, run):
    filelist = sorted(glob.glob('/sf/alvra/data/{}/res/processed/run{:04d}*/data/*acq*corr*'.format(pgroup, run)))
    return filelist

def load_one_acq(pgroup, run):
    file2open = generate_filelistJF_corr(pgroup, run)[0]
    if not file2open:
        print ('===> run{:04d} not found in res/processed/ folder! <==='.format(run))
        return None, None, None
    else: 
        result = {}
        with h5py.File(file2open, 'r') as d:
            for group_name in d.keys():
                group = d[group_name]
                result[group_name] = {}
                
                for dataset_name in group.keys():
                    result[group_name][dataset_name] = group[dataset_name][()]

        return result

def load_step_scan_roi(pgroup, run, quantile, pp=False):
    filelist = generate_filelistJF(pgroup, run)
    if pp: 
        filelist = generate_filelistJF_pp(pgroup, run)
    if not filelist:
        print ('===> run{:04d} not found in res/processed/ folder! <==='.format(run))
        return None, None, None
    else: 
        readbacks = []
        intensity = []
        for i,step in enumerate(tqdm(filelist)):
            with h5py.File(step, 'r') as dd:
                readbacks.extend(dd['meta']['readback_value'][:])
                df = pd.DataFrame(dd['data']['intensity'])
                intensity.append(np.nanquantile(df, [0.5, 0.5 - quantile/2, 0.5 + quantile/2]))
    return np.asarray(intensity), np.array(readbacks)

def unwrap_data2(data):
    print ("Loaded {} steps".format(len(data)))
    signal   = data[:,0]
    err_low  = data[:,1]
    err_high = data[:,2]    
    return signal, err_low, err_high
```

```python
def load_step_scan_radial(pgroup, run):
    filelist = generate_filelistJF_pp(pgroup, run)
    if not filelist:
        print ('===> run{:04d} not found in res/processed/ folder! <==='.format(run))
        return None, None, None
    else: 
        readbacks = []
        intensity = []
        rp_pump   = []
        rp_unpump = []
        rp_pumpprobe = []
        for i,step in enumerate(tqdm(filelist)):
            with h5py.File(step, 'r') as dd:
                readbacks.extend(dd['meta']['readback_value'][:])
                rp_pump.append(dd['data']['rp_pump'][:])
                rp_unpump.append(dd['data']['rp_unpump'][:])
                rp_pumpprobe.append(dd['data']['rp_pp_avg'][:])
                #rp_pumpprobe.append(rpp)
                intensity.append(dd['data']['intensity'][()])
    return np.asarray(intensity), np.array(readbacks), np.array(rp_pump), np.array(rp_unpump), np.array(rp_pumpprobe)
```

```python
class XANES_analysis():
    def __init__(self, pgroup, runlist, withTT=False, t0_offset=None):
        if t0_offset == None:
            t0_offset = [0]*len(runlist)
        self.pgroup = pgroup
        self.runlist = runlist
        self.withTT = withTT
        self.t0_offset = t0_offset

        runname = []
        for run in self.runlist:
            if glob.glob('/sf/alvra/data/{}/res/processed/run{:04d}*'.format(pgroup, run)):
                runname.append(run)
            else:
                print ("Could not find run {} reduced in pgroup {}".format(run, pgroup))

        print ("will load run(s): {}".format(runname))
        self.runlist = runname

    def merge_steps(self):
        from collections import defaultdict
        from sfdata import SFScanInfo
        for run in self.runlist:
            jsonfile = get_jsonfile_from_run(run, self.pgroup)
            scan = SFScanInfo(jsonfile)
            self.readbacks = scan.readbacks
            filelist = sorted(glob.glob('/sf/alvra/data/{}/res/processed/run{:04d}*/data/*acq*'.format(self.pgroup, run)))
            d = defaultdict(list)
            for i, acq in enumerate(filelist):
                with h5py.File(acq, 'r') as dd:
                    for key, value in dd.items():
                        d[key].extend(value)
            d["readbacks"] = self.readbacks
            return d

    def knife_edges(pgroup, run, channel_list, quantile):
        jsonfile = get_jsonfile_from_run(run, pgroup)
        scan = SFScanInfo(jsonfile)
        readbacks = scan.readbacks
        Signal    = []
        ch_sig   = channel_list[0]
        ch_izero = channel_list[1]
        filelist = generate_filelist(pgroup, run) 
        for i,step in enumerate(filelist):
            with h5py.File(step, 'r') as dd:
                df = pd.DataFrame(np.array(dd[ch_sig])/np.array(dd[ch_izero]))
                Signal.append(np.nanquantile(df, [0.5, 0.5 - quantile/2, 0.5 + quantile/2]))
        return np.asarray(Signal), readbacks
```

```python
def timestamp(json_file):
    file_split = json_file.split('/')[:-1]
    path_to_bsdata = '/'.join([*file_split[:-1], 'data', '*BSDATA.h5'])
    
    timestamp_s = []
    
    for file in glob.glob(path_to_bsdata):
        with h5py.File(file) as f:
            timestamp_ns = f[channel_Events]['timestamp'][:]
            timestamp_s.append(np.mean(timestamp_ns) * 1e-9)
    timestamp_s = np.mean(timestamp_s)
    timestamp_datetime = datetime.fromtimestamp(timestamp_s)
    return np.datetime64(timestamp_datetime)

def timestamp_step(step):
    bsdata = step.fnames[0]
    timestamp_s = []
    with h5py.File(bsdata) as f:
        timestamp_ns = f[channel_Events]['timestamp'][:]
        timestamp_s.append(np.mean(timestamp_ns) * 1e-9)
    timestamp_s = np.mean(timestamp_s)
    timestamp_datetime = datetime.fromtimestamp(timestamp_s)
    return np.datetime64(timestamp_datetime)
```
